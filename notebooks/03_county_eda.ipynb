```python
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (11, 5)

# Ganesh's county-level static file (rainfall climatology + SVI themes)
cs = pd.read_csv("C:\\Users\\SUDARSHAN\\climate-risk-dashboard\\data\\processed\\county_static.csv")

print("Shape:", cs.shape)
print("Columns:", list(cs.columns))
cs.head()
```

```python
# Extreme-rainfall exposure (p95 3-day) vs social vulnerability
plt.figure(figsize=(9, 6))
plt.scatter(cs["RPL_THEMES"], cs["p95_fwd3"], s=cs["pop_density"]/15,
            alpha=0.5, color="steelblue", edgecolor="white", linewidth=0.5)
plt.xlabel("Social Vulnerability (SVI percentile — higher = more vulnerable)")
plt.ylabel("Extreme 3-day rainfall threshold (p95, mm)")
plt.title("Rainfall exposure vs social vulnerability across 356 Gulf counties")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Quick correlation
print("Correlation (vulnerability vs extreme rainfall):",
      round(cs["RPL_THEMES"].corr(cs["p95_fwd3"]), 3))
```

```python
plt.figure(figsize=(12, 6))
sc = plt.scatter(cs["lon"], cs["lat"], c=cs["RPL_THEMES"], cmap="YlOrRd",
                 s=cs["pop_density"]/12, alpha=0.8, edgecolor="gray", linewidth=0.3)
plt.colorbar(sc, label="Social Vulnerability (SVI percentile)")
plt.xlabel("Longitude"); plt.ylabel("Latitude")
plt.title("Social vulnerability across Gulf Coast counties (dot size = pop. density)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("../outputs/figures/v2_fig2_svi_map.png", dpi=150, bbox_inches="tight")
plt.show()
```
